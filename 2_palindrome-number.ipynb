```python
class Solution:
    def isPalindrome(self, x: int) -> bool:
        digits_list = list(str(x))
        digits_list.reverse()
        reversed = ''.join(digits_list)
        print(reversed)
        print(x)
        if(str(x) == reversed):
            return True
        else:
            return False
```

```python
solution = Solution()
solution.isPalindrome(10301)
```
