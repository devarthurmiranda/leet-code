```python
class Solution:
    def longestCommonPrefix(self, strs: list[str]) -> str:
        if not strs:
            return ""

        shortest = min(strs, key=len)

        for i, char in enumerate(shortest):
            for word in strs:
                if word[i] != char: # if character does not match the next word
                    print('word: ' + word[i])
                    print('char: ' + char)
                    print('new shortest: ' + shortest[:i])
                    return shortest[:i]

        return shortest
```

```python
solution = Solution()
solution.longestCommonPrefix(['test', 'teste', 'testo', 'teava'])
```
