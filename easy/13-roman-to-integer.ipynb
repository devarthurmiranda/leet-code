```python
class Solution:
    def romanToInt(self, s: str) -> int:
        values = {'I': 1, 'V': 5, 'X': 10, 'L': 50, 'C': 100, 'D': 500, 'M': 1000}
        result = 0

        for i, char in enumerate(s):
            result += values[char]
            if i > 0 and values[s[i-1]] < values[char]:
                result -= 2 * values[s[i-1]]

        return result
```

```python
solution = Solution()
solution.romanToInt('IVXC')
```
